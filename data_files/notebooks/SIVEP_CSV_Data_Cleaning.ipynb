```python
import numpy as np
import pandas as pd
import glob
import os
import regex as re
```

```python
# Create output folder if it doesn't exist
os.makedirs(output_folder, exist_ok=True)
```

```python
# Columns to drop
columns_to_drop = ['Unnamed: 0', 'DT_TRATA', 'ESQUEMA', 'CO_FORMA_SANGUINEA', 
                   'CO_FORMA_TECIDUAL_GAMETOCITO', 'EXAME ', 'NU_ATIVIDADE_G6PD', 
                   'NU_HEMOGLOBINA_TOTAL', 'EXAMINADOR', 'ID_LVC ', 'GESTANTE ', 
                   'AMAMENTANDO', 'TEMP_AMAMENT', 'DIMEA_AMAMENT', 'NIV_ESCO', 'FALCIPARUM',
                   'QTD_PARA', 'QTD_CRUZ', 'SEXO', 'RACA', 'PESO', 'COD_AGEN', 'DT_NASCI', 'ID_PACIE',
                   'ID_DIMEA', 'COD_UNIN', 'COD_NOTI', 'TIPO_LAM', 'DT_ENVLO', 'HEMOPARASI', 'DT_DIGIT', 'VIVAX']
```

```python
# Process each year from 2003 to 2023
for year in range(2003, 2024):
    year_suffix = str(year)[-2:]
    input_path = f"../../data_files/data/sivep_notification_data/malaria_notification_data_csv/NOTIPO{year_suffix}.csv"
    
    # Read and process
    df = pd.read_csv(input_path, low_memory=False)
    df = df.drop(columns=columns_to_drop, errors='ignore')
    df = df.dropna(axis=1, how='all')
    df = df[df['RES_EXAM'] == 4]
    df = df[df['PAIS_INF'] == 1]
    df = df[df['PAIS_RES'] == 1]
    df = df.drop(columns=['RES_EXAM', 'PAIS_INF', 'PAIS_RES'])
    
    # Create named dataframe
    globals()[f"malaria_{year}_df"] = df
    
    # Save to CSV
    output_path = f"../../data_files/data/sivep_notification_data/treated_malaria_notification_data/malaria_vivax_{year}_treated.csv"
    df.to_csv(output_path, index=False)
    
    print(f"✓ Created malaria_{year}_df with {len(df):,} rows")
```

```python
# Collect all yearly dataframes
dfs = [globals()[f"malaria_{year}_df"] for year in range(2003, 2024)]

# Combine into a single dataframe
combined_df = pd.concat(dfs, ignore_index=True)

# Ensure datetime
combined_df['DT_SINTO'] = pd.to_datetime(combined_df['DT_SINTO'], errors='coerce')
combined_df['DT_NOTIF'] = pd.to_datetime(combined_df['DT_NOTIF'], errors='coerce')
combined_df['DT_EXAME'] = pd.to_datetime(combined_df['DT_EXAME'], errors='coerce')

combined_df
```

```python
# In some cases, there were multiple years between notification and detection of symptoms (e.g.: 'malaria_2003_df.iloc[24428]')

# To avoid these outliers, and knowing that P. vivax may remain inactive in the organism up to at most 5 years, we will limit the dfference
# between DT_NOTIF and DT_SINTO to 1830 days at most

# --------------------------------------------------
# DATE CONSISTENCY FILTERS
# --------------------------------------------------

# Differences in days
diff_sinto_notif = (combined_df['DT_SINTO'] - combined_df['DT_NOTIF']).dt.days
diff_exam_notif  = (combined_df['DT_EXAME'] - combined_df['DT_NOTIF']).dt.days
diff_exam_sinto  = (combined_df['DT_EXAME'] - combined_df['DT_SINTO']).dt.days

# 1) Limit biologically plausible relapse window
# P. vivax hypnozoites rarely exceed ~5 years
valid_relapse_window = (diff_sinto_notif <= 1830) | (diff_sinto_notif.isna())

# 2) Symptoms should not occur many days AFTER notification
# Notification normally happens close to exam/symptoms
valid_notification_timing = (diff_sinto_notif <= 30) | (diff_sinto_notif.isna())

# 3) Exam should not occur long before notification
valid_exam_notification = (diff_exam_notif >= -7) & (diff_exam_notif <= 30)

# 4) Exam should occur near symptoms
valid_exam_symptom = (
    (diff_exam_sinto >= -30) &
    (diff_exam_sinto <= 30)
) | (diff_exam_sinto.isna())

# Apply filters
clean_df = combined_df[
    valid_relapse_window &
    valid_notification_timing &
    valid_exam_notification &
    valid_exam_symptom
].copy()

clean_df = clean_df.reset_index(drop=True)

clean_df
```

```python
# Asymptomatic patients
asymptomatic_df = clean_df[clean_df['SINTOMAS'] == 2]

# Check inconsistent rows:
# asymptomatic but symptom date exists
inconsistent_asymptomatic = asymptomatic_df[
    asymptomatic_df['DT_SINTO'].notna()
]

print("Number of inconsistent asymptomatic records:")
print(len(inconsistent_asymptomatic))

inconsistent_asymptomatic
```
