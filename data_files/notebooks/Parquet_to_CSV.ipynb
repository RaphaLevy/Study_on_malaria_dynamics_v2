```python
import pandas as pd
import glob
import os
```

```python
input_folder = "../../data_files/data/sivep_notification_data/malaria_notification_data/"
output_folder = "../../data_files/data/sivep_notification_data/malaria_notification_data_csv/"
```

```python
# Ensure the output folder exists
os.makedirs(output_folder, exist_ok=True)

# Get all parquet files
parquet_files = glob.glob(os.path.join(input_folder, "*.parquet"))
parquet_files
```

```python
# Convert each parquet file to CSV
for file in parquet_files:
    df = pd.read_parquet(file)
    filename = os.path.basename(file).replace(".parquet", ".csv")
    output_path = os.path.join(output_folder, filename)
    df.to_csv(output_path, index=False)
    print(f"Saved: {output_path}")

print("Conversion completed!")
```
